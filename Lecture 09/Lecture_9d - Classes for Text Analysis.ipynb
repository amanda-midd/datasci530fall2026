# <span style="color:blue"> I. Store Text and Count a Word </span>

<font size = "5">

Create a class with one attribute and one function

- ```__init__``` stores a sentence in the attribute ```text```
- ```count_word()``` counts how many times a word appears
- ```self.text``` accesses the text belonging to that instance

</font>

````python
class text_template:

    def __init__(self, text_arg):
        self.text = text_arg

    def count_word(self, word_arg):
        # Convert to lowercase and split the sentence into words
        words = self.text.lower().split()
        return words.count(word_arg.lower())
````

<font size = "5">

Create two instances with different sentences

Both instances use the same function, but each stores its own text.

</font>

````python
text1 = text_template("The Senate debated the bill")
text2 = text_template("The House debated the bill and passed the bill")

print(text1.text)
print(text2.text)
````

<font size = "5">

Apply the function to each instance

We supply the word to count. Python supplies ```self``` automatically.

</font>

````python
print(text1.count_word("bill"))
print(text2.count_word("bill"))
````

<font size = "5">

Change the text attribute and count again

</font>

````python
text1.text = "The bill is a Senate bill"
text1.count_word("bill")
````

<font size = "5">

This simple function splits text at spaces. <br>
It counts ```"bill"```, but not ```"bill."``` with a period. <br>
For now, use sentences without punctuation.
