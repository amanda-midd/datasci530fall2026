# <span style="color:darkblue"> II. Basics of Formatting </span>

<font size = "5">

Inserting variables inside strings with "fstrings"


```python
# Define a string
dayofweek = "Thursday"
# An "fstring" always starts with the letter 'f'
# Followed by a section with double quotations
# Inside you write your desired string

f"Today is {dayofweek}"
```

```python
# Another example
month = 8
day   = 28

f"Example of month-day fomatting: {month}-{day}"
```

<font size = "5">

Formatting codes control how a number is displayed.

- `.0f` displays a whole number
- `.2f` displays two digits after the decimal point


```python
# The letter "f" means fixed-point notation.
example_number  = 14.182258

print(f"{example_number : .0f}")
print(f"{example_number : .2f}")

# Note that the inner argument is a dictionary
# telling Python to assign format ".0f" to they key "example_number"
```

```python
# Other formatting codes can be combined with .0f or .2f.
large_number = 12500.5

print(f"{large_number:,.0f}")
print(f"{large_number:,.2f}")
print(f"${large_number:,.2f}")
```

# <span style="color:darkblue"> III. Format a table </span>

<font size = "5">

Create an example table to work with.

```python
# Apply formatting directly to a table for sharing.
# The original DataFrame remains unchanged.
example_costs = pd.DataFrame({
    "category": ["Entry fee", "Equipment"],
    "average_cost": [1250.5, 875.0],
    "obs": [4, 12]
})

example_costs

# The inner argument is a dictionary
# which gets converted to a dataframe to be in table format.
```

<font size = "5">

Format the entries

```python
# Format the table
# Three commands are being applied one after the other to the object "example_costs"
# (i) .style.format() modified the formatting of certain values,
#     taking a dictionary {"average_cost": ... , "obs": ...} as an input
# (ii) .hide() removes the row index column
# (iii) 

newtable_1 = example_costs.style.format({
    "average_cost": "${:,.2f}",
    "obs": "{:,.0f}"
}).hide(axis="index").set_caption("Cost summary")
```

<font size = "5">

Adjust column names

```python
# Rename columns with a dictionary.
newtable_2 = example_costs.rename(columns={
    "Category": "Cost category",
    "Average cost": "Mean cost",
    "Observations": "Number of observations"
})

example_costs
```

# <span style="color:darkblue"> III. Format a descriptive statistics table </span>

<font size = "5">

Create a raw table of descriptive statistics

```python
# Named aggregations become columns in the resulting DataFrame.
# groupby("driverId") creates one row for each driver.
drivers_agg = (results.agg(mean_points = ("points", "mean"),
                           sd_points   = ("points", "std"),
                           min_points  = ("points", "min"),
                           max_points  = ("points", "max"),
                           count_obs   = ("points", len)))

drivers_agg.head()
```

<font size = "5">

Round entries

```python
# Use the same format codes in a shareable DataFrame table.
drivers_agg.style.format({
    "points": "{:,.1f}"
}).set_caption("Driver performance summary")
```
