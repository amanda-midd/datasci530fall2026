## Gradient Descent / stochastic gradient descent:

\* an optimization algorithm used to find model parameters—weights and biases—that minimize a loss function


### SGD
- updates the parameters using one randomly selected training observation at a time (to minimize running time)
- SGD updates are faster but noisier but noise can also help the model escape poor local solutions

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta \nabla L(\theta)
$$


$(\theta): model parameters $ <br>
$(L): loss function $ <br>
$(\nabla L): gradient of the loss$ <br>
$(\eta): learning rate$ 


the learning rate controls the update size
- Too small: training is slow.
- Too large: training may overshoot or fail to converge

## Principal components analysis (PCA):
$\rightarrow$ reduces features/variable into a lower dimension space while retaining as much variation as possible

- Creates new variables called principal components.
    - Variables should usually be standardized before PCA when they use different scales
- Each component is a weighted combination of the original variables.
- Components are uncorrelated with one another



\* reality: no data can contain all variation in 1 variation <br>
can ideally cover MOST variation with first 2 principal components, but can do as many as is necessary


## K-means Clustering
divides observations into \(K\) clusters based on **similarity**. usually meant for *classification* <br>
\* works best with roughly spherical, similarly sized clusters

We choose:
- number of clusters, *k*

basic process:
1. Initialize \(K\) cluster centers, called **centroids**
    - usually need to standardize variables before clustering
    - centroids represent basic cluster areas, it isn't a point itself
        - Results can depend on the initial centroids $\leftarrow$ begins pretty much randomly. then iteratively makes better clusters
2. Assign each observation to its nearest centroid.
3. Recalculate each centroid using the assigned observations.
4. Repeat until assignments stop changing.



- methods such as the **elbow method** or **silhouette score** can help select best \(K\)
K-means minimizes the within-cluster sum of squared distances:

# <span style="color:blue"> III. Principal Components Analysis </span>


<font size = "5">

A factor is a linear combination of the columns of $X$

$F = X \omega$

We call $\omega \in \mathbb{R}^K$ the factor loadings.

- More factors can describe the data better
- Fewer factors reduce dimensionality, i.e. regularize.
- Principal Components Analysis (PCA) is an algorithm <br>
to obtain factors.


<font size = "5">

Intuition: The first factor solves

$ \quad \omega_1 = \arg\max_{\omega \in \mathbb{R}^K} \frac{\omega^TX^TX\omega}{\omega^T\omega} $

$F_i^{FIRST} = \sum_{k=1}^K \omega_k^{FIRST} X_{ik}$

- This tries to compute the most variation in $X$
- The second factor solves a similar problem, with the $X$ residual <br>
not explained by the first factor

````python
# Choose number of components

pca = PCA(n_components=3)
# fit PCA model to beast cancer data
pca.fit(X_train)
````

<font size = "5">

We can check the loadings for each factor

````python
# pca_components_ has the factor loadings
factor_loadings = pd.DataFrame(pca.components_,
                               columns = cancer.feature_names)

factor_loadings
````

<font size = "5">

A way to visualize loadings

````python
plt.matshow(pca.components_, cmap='viridis')
plt.yticks([0,  1], ["First component", "Second component"])
plt.colorbar()
plt.xticks(range(len(cancer.feature_names)),
           cancer.feature_names, rotation=60, ha='left')
plt.xlabel("Feature")
plt.ylabel("Principal components")
````

<font size = "5">

Compute factors

````python
# transform data onto the first two principal components

X_pca = pca.transform(X_train)
````

<font size = "5">

Are factors useful for prediction?

````python
# Hue is used to color

ax = sns.scatterplot(x = X_pca[:,0],
                     y = X_pca[:,1],
                     hue = y_train)
ax.set(xlabel='First Principal Component',
       ylabel='Second Principal Component')
ax.legend(title='Cancer Screening', loc='upper left', labels=
                      ["Benign","Malignant"])
````

# <span style="color:blue"> IV. K-Means Clustering </span>


<font size = "5">

K-means clsutering is a closely related algorithm for clustering <br>
groups of observations


````python
# ncentroids is the number of cluster
# niter controls how many iterations. "KMeans" is an iterative algorithm,
# the more iterations yuo run, the close you are the optimum
# d is the number of columns
# verbose controls where you get messages with the status of the process

ncentroids = 3
niter = 100
verbose = True
d = X_train.shape[1]

kmeans = faiss.Kmeans(d,
                      ncentroids,
                      niter=niter,
                      verbose=verbose,
                      seed = 0)
````

<font size = "5">

Train and compute clusters

````python
# Convert to float 
Xtilde = np.float32(X_train)
kmeans.train(Xtilde)

# Assign observations to clusters
results = kmeans.index.search(X_train, 1)

# This computes the cluster membership
indexes_train = results[1]
#indexes_train
````
