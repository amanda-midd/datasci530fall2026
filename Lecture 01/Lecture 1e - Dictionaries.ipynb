In the previous lecture we covered lists to store objects.

- An alternative are dictionaries
- Lists retrieve information based on a position {0,1,2,3}, 
whereas dictionaries use "keys" (named values) 
- Both allow for nesting
- They can be combined

<font size = "5">

Examples of Dictionary Formatting

- Each "key" is associated with a "value"
- Different keys can store different data types

```python
# Define dictionary
dictionary_student = \
{
    "first_name": "John",
    "last_name" : "Smith",
    "location"  : "Atlanta",
    "university": "Emory",
    "age": 30
}

# Retrieve specific information
# key: "first_name"
# value: "John"

print( dictionary_student["first_name"])
```

<font size = "5">
Nested dictionaries


```python
dictionary_student_wgrades =\
{
    "first_name": "John",
    "last_name" : "Smith",
    "location"  : "Atlanta",
    "university": "Emory",
    "grades":{
        "computing":"A",
        "reasoning":"B",
        "applied":"A-"
    }
}

# Retrieve information within the list

print(dictionary_student_wgrades["grades"])
print(f"\n pulling information from nested dictionary/list \n computing class: {dictionary_student_wgrades["grades"]["computing"]}")
```

<font size = "5">
Nested dictionaries + Lists

```python

# 1 list containing 2 directories 

dictionary_liststudents = \
[{
    "first_name": "Ned",
    "last_name" : "Stark",
    "location"  : "Atlanta",
    "university": "Emory",
    
 },
 {
    "first_name": "Cersei",
    "last_name" : "Lannister",
    "location"  : "Decatur",
    "university": "Georgia Tech",
 }
]

#pulling from 1st component of list and calling for the variable "first_name"
print( dictionary_liststudents[0]["first_name"]) 
```
